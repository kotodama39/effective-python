# for ループと while ループの後の else ブロックは使わない

Python のループには、他のほとんどのプログラミング言語にはない特別な機能があります。

```python
for i in range(3):
  print(i)
else:
  print("Done")
```

else ブロックは、ループが終了した直後に実行されます。

if / else 文では、else は「この前のブロックがじっこうされなかったら、これを実行しなさい」という意味です。

が、for ループの後の else は異なった動作をします。

これは読みづらいです。

ちなみにループで break 文が実行されると、実は else ブロックがスキップされます。

```python
for i in range(3):
  print(i)
  if i == 1:
    break
else:
  print("Done")
```

さらに驚くべきは、空シーケンスでループしたら、else ブロックがすぐ実行されることです。

```python
for x in []:
  print(x)
else:
  print("Done")
```

else ブロックは、while ループが頭で失敗したときにも実行されます。

```python
while False:
  print("Loop body")
else:
  print("Done")
```

else ブロックによる表現は、（自分も含めた）人が将来そのコードを理解するのに必要な負荷にあたいしません。

## 覚えておくこと

- Python には、for や while のループブロックの直後に else ブロックを許す特別な構文がある。
- ループの後の else ブロックは、ループ本体で break 文が実行されなかった場合にのみ実行される。
- ループ直後の else ブロックは、振る舞いが直感的でなく、誤解を生みやすいので使用しない。
