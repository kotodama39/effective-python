# インデックスではなく複数代入アンパックを使う

Python 組み込みのタプルは、値の変更不能な順序シーケンスを作る方です。単純な形式のタプルの例は、辞書キーと値からなる 2つの値の対です。

```python
snack_calories = {
  'chips': 140,
  'popcorn': 80,
  'nuts': 190,
}
item = tuple(snack_calories.items())
print(f'Item: {item!r}')
```

タプルの値は、インデックスでアクセスできます。

```python
item = ('Peanut butter', 'Jelly')
first = item[0]
second = item[1]
print(f'First: {first!r}, Second: {second!r}')
```

作られたタプルに新たな値をインデックスで代入することはできません。

```python
pair = ('Peanut butter', 'Jelly')
pair [0] = 'Banana'  # エラー
```

Python には、アンパック構文があり、1つの代入分で複数の値に代入できます。

```python
item = ('Peanut butter', 'Jelly')
first, second = item
print(f'First: {first!r}, Second: {second!r}')
```

アンパックは、インデックスでタプルにアクセスする場合よりも見た目がすっきりして、行数が少なくなります。

これと同じパターンの構文をリスト、シーケンス、イテラブル内の深いレベルにあるイテラブルへの代入にも使えます。

次のようなコードを進めるわけではありｍせんが、これが可能であり、どのように動作するかを知っておくことは重要です。

```python
favarite_snacks = {
  'salty': ('pretzels', 100),
  'sweet': ('cookies', 180),
  'veggie': ('carrots', 20),
}

((type1, (name1, cals1)),
 (type2, (name2, cals2)),
 (type3, (name3, cals3))) = favarite_snacks.items()

print(f'{type1.title()} snack: {name1} has {cals1} calories')
print(f'{type2.title()} snack: {name2} has {cals2} calories')
print(f'{type3.title()} snack: {name3} has {cals3} calories')
```

重要なアンパックの用途が、forループや同様の内包表記とジェネレータ式（項目27）におけるターゲットリストです。

```python
# 比較対象用のアンパックを使わないコード例
# スナック食品リストのイテレーション例
snacks = [('bacon', 350), ('donut', 240), ('muffin', 190)]
for i in range(len(snacks)):
    snack = snacks[i]
    name = snack[0]
    calories = snack[1]
    print(f'Snack: {name}, Calories: {calories}')
```

これは動きますが、とても読みづらいコードです。

snacks の構造の深さごとにインデックスを使うため文字が余計に必要です。

アンパックを組み込み関数 enumerate (項目7) とともにつかうことで同じ出力結果が次のコードで得られます。

```python
for rank, (name, calories) in enumerate(snacks, 1):
    print(f'{rank}. Snack: {name}, Calories: {calories}')
```

アンパックを賢く使えば、可能ならインデックスを使わずに済み、より明確で Pythonic なこー度になります。

## 覚えておくこと

- Python には1つの代入分で複数の値を代入できるアンパック構文がある。
- アンパックは Python では一般的であり、イテラブル内のレベルにあるイテラブルを含めて、どのようなイテラブルにも使える。
- アンパックを使えば、シーケンスにインデックスを使わずに済むので、コードが明確になり、見た目がすっきりする。

## Tips

\* 演算子とアンパックを組み合わせると以下のようなことができます。

```python
# 例えば、先頭だけを取り出したいとき
first, *rest = [1, 2, 3, 4, 5]
print(f'First: {first}, Rest: {rest}')
```

```python
# 末尾だけを取りたいとき
*start, last = [1, 2, 3, 4, 5]
print(f'Start: {start}, Last: {last}')
```
