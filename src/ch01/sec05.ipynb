# 複雑な式の代わりにヘルパー関数を書く

Python の簡潔な構文を使えば、多数のロジックを実装した１行の式を簡単に書くことができます。

```python
# 例： URL のクエリ文字列を腹腔がしたいとする
# クエリ文字列引数は、次のように整数値であらわされる
from urllib.parse import parse_qs
query_string = 'red=5&blue=0&green='
my_values = parse_qs(query_string, keep_blank_values=True)
print(params)
```

```python
# 存在するが空白値を持つもの、あるいは欠損値に対して異なる値を返す
print('Red: ', my_values.get('red'))
print('Green: ', my_values.get('green'))
print('Opacity: ', my_values.get('opacity'))
```

```python
# red の場合、キーが my_values に存在するため、この文字列は暗黙的に True に評価されます。
red = my_values.get('red', [''])[0] or 0
# green の場合、キーは存在しますが、空白値を持つため、この文字列は暗黙的に False に評価されます。
green = my_values.get('green', [''])[0] or 0
# opacity の場合、キーが存在しないため、デフォルト値の空文字列が返され、これは暗黙的に False に評価されます。
opacity = my_values.get('opacity', [''])[0] or 0
print(f'Red: {red!r}')
print(f'Green: {green!r}')
print(f'Opacity: {opacity!r}')
```

```python
green_first = my_values.get('green', [''])[0]
print(f'Get green first value is {green_first!r}')
```

```python
# しかしこれは全ての引数値が整数になって、数式に使えることを保証するためには、
# 各式を組み込み関数 int でラップして文字列を整数としてパースします
red = int(my_values.get('red', ['0'])[0] or 0)
print(f'Red: {red!r}')
```

これは非常に読みにくい

読み手は、何をしているのかを理解するために式を分解して理解する必要があります。

短くすることはいいことですが、これらすべてを1行に収めて書く価値はありません。

Python では、if / else 条件式、すなわち三項式で、このような場合にコードを短く保ちながら、より明確にすることが可能です。

```python
red_str = my_values.get('red', [''])
red = int(red_str[0]) if red_str[0] else 0
print(f'Red: {red!r}')
```

あまり複雑でない状況では、if/else条件文は非常に明確です。

が、この例の場合は、複数行にまたがる完全な if/else文ほどに明確ではありません。

```python
green_str = my_values.get('green', [''])
if green_str[0]:
    green = int(green_str[0])
else:
    green = 0
print(f'Green: {green!r}')
```

このロジックを何回か使う必要があるなら、ヘルパー関数を書くことが解決法になります。

```python
def get_first_int(values, key, default=0):
    str_value = values.get(key, [''])
    if str_value[0]:
        return int(str_value)
    else:
        return default
```

呼び出しコードは、or を使った複合式あ if / else 条件式を使った 2行のものよりもずっと明確になります。

```python
green = get_first_int(my_values, 'green')
print(f'Green: {green!r}')
```

## 覚えておくこと

- Python の構文では、ただ複雑なだけで読みにくい１行の式が簡単に書けてしまう。
- 複雑な式（特に、同じロジックを繰り返す必要がある場合）は、ヘルパー関数にする。
- if / else 条件式は、or や and のような論理演算子を使用するよりも読みやすい代替手段を提供する
