# PEP8 スタイルガイドに従う

https://pep8-ja.readthedocs.io/ja/latest/

## 名前付け

* 関数、変数、属性は、lowercase_underscore
* protected 属性は、_leading_underscore
* private 属性は、__double_underscore
* クラスと例外は、CapitalizedWord
* モジュールでの定義は、ALL_CAPS
* クラスのインスタンスメソッドは、第一パラメータの名前に self
* クラスメソッドは、第一パラメータの名前に cls 

## 式と文

* if not a is b ではなく、if a is not b
* コンテナやシーケンスの長さ（if len(somelist) == 0）を使って、空値（[] や '' など）かどうかをチェックしない。if not somelist を使って、空値が暗黙に False と評価されることを使う
* 上と同じで、非空値（[1] や 'hi'）にも使う。非空値について、if somelist は、暗黙に True と評価される
