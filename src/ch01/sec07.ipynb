# range ではなく enumerate を使う

組み込み関数 range は、整数集合上でループする繰り返し処理で役立ちます。

```python
from random import randint

random_bits = 0
for i in range(64):
  if randint(0, 1):
    random_bits |= 1 << i

print(bin(random_bits))
```

文字列のリストのようなデータ構造では、シーケンスを直接ループできます。

```python
flavor_list = ['vanilla', 'chocolate', 'pecan', 'strawberry']
for flavor in flavor_list:
    print(f'{flavor} is delicious')
```

リストの処理で、リスト中の要素のインデックスが必要な際に、range を使うこともできます。

```python
for i in range(len(flavor_list)):
    print(f'{i + 1}: {flavor_list[i]}')
```

ただしこれはリストの長さや配列のようにインデックスを使う必要があります。

そのため、ステップが多くて読むのが面倒です。

## enumerate 関数

Python には、このような状況に対する組み込み関数 enumerate があります。

enumerate は、遅延評価ジェネレータ（項目30）でイテレータをラップします。

enumerate は、ループのインデックスとイテレータの次の値の対を yield します。

```python
# 戻り値のイテレータに組み込み関数 next を使って、次の値を求めています。
it = enumerate(flavor_list)
print(next(it))  # (0, 'vanilla')
print(next(it))  # (1, 'chocolate')
```

enumerate で yield される各対は、for 文でうまくアンパックできます。

```python
for i, flavor in enumerate(flavor_list):
    print(f'{i + 1}: {flavor}')
```

```python
# enumerate でカウントを開始する数を指定すれば、さらにみやすくなります。
for i, flavor in enumerate(flavor_list, 1):
    print(f'{i}: {flavor}')
```

## 覚えておくこと

- enumerate の簡潔な構文で、イテレータでループしながら、要素のインデックスを取り出すことができる。
- range でループして、シーケンスのインデックスを処理するよりも、enumerate を使う方がよい。
- enumerate の第２引数で、カウンタを開始する数（デフォルトはゼロ）を指定できる
